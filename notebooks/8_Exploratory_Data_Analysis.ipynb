```python
df_features = df.iloc[:,2:]
df_features.head(5)
```

```python
feature_col = df_features.columns
```

```python
is_binary = (df_features.nunique() == 2) & (df_features.min() == 0) & (df_features.max() == 1)
zero_counts = (df_features == 0).sum()
```

```python
# One-hot encoded cols
binary_cols = feature_col[is_binary]

# These columns have values like 0 power, 0 sccm, or 0% as they were not present in that specific sample (exclusing one-hot encoded)
zero_count_columns = feature_col[(zero_counts > 0) & (~is_binary)] # Might have to adjust 0

# Columns with full values
filled_columns = feature_col[zero_counts == 0]
```

```python
# From this, the columns that are supposed to be binary, contain values of 0 at some points, and contain completely filled data are confirmed to be as intended

# Gather columns without xrd measurments
features_without_xrd_columns = [col for col in feature_col if "xrd" not in col]
df_featuers_without_xrd = df[features_without_xrd_columns]
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Make a heatmap plot
corr_matrix = df_featuers_without_xrd.corr()
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))

plt.figure(figsize=(24, 16))
sns.heatmap(corr_matrix,
             annot=False,
             fmt='.2f',
             cmap="coolwarm",
             cbar=True,
             mask=mask,
             vmin=-1,
             vmax=1)
plt.title("Correlation heatmap of features (excluding xrd)")
plt.show()
```

```python
# Heatmap only contains trivial relationships:
# 1.) If a certain deposition compound was used (ex: InO), the material would contain compounds with corressponding elements (In2O3)
# 2.) If a substrate material is present, another is not (as only one substrate material will be present at a time)

# Concentrate on treds from other deposition parameters + thickness
unique_features_col = []
exclude = [
    "xrd",      # xrd bg and intensity values
    "xrf",      # xrf compounds
    "power",    # deposition compounds
    "sccm",     # deposition gases
    "substrate" # deposition substrate
]

for col in feature_col:
    for val in exclude:
        if val in col: 
            break
    else:   
        unique_features_col.append(col)
# df_unique_features = df_features[unique_features_col]
```

```python
unique_features_col
```

```python
import math
```

```python
# Currently, 'thickness' seems to be the most likely target value for a Neural Net
# Create a histrogram for these features + the target feature ('thickness')
items_per_row = 2
fig, axes = plt.subplots(math.ceil(len(unique_features_col)/items_per_row), items_per_row, figsize=(15, len(unique_features_col)*4))
axes = axes.flatten()

for i, col in enumerate(unique_features_col):
    skew_value = df[col].skew()
    sns.histplot(df[col], kde=True, bins=45, ax=axes[i])
    axes[i].set_title(f"Histogram of {col} (Skew: {skew_value:.2f})")
    axes[i].set_xlabel(col)
    axes[i].set_ylabel("Frequency")

plt.subplots_adjust(hspace=0.3, wspace=0.2) 
plt.show()
```

```python
# Both 'thickness' and 'base pressure' have the least apparent variance

# Find features in general that sem to correlate (positively or negatively) with 'thickness'
thickness_corr = df_features.corr()

top_n = 20
top_features = (
    thickness_corr['thickness']
    .abs()
    .sort_values(ascending=False)
    .head(top_n+1)
    .index
)

plt.figure(figsize=(24, 16))
sns.heatmap(
    thickness_corr.loc[top_features, ['thickness']]
        .sort_values(by='thickness', ascending=False),
    fmt='.2f',
    cmap="coolwarm",
    cbar=True,
    vmin=-1,
    vmax=1,
    annot=True
)

plt.title(f"Correlation of top {top_n} descriptors with thickness")
plt.show()
```

```python
# judging by the near 0 values of the three xrd in the list
# it appears that there is not one dominant xrd angle that matters for thickness
# which makes sense as the samples are part of differing material systems

# Tantanium being correlated with thickness makes sense conceptually as it is a bigger atom
# Correlation disaproportionally weighs each parameter that contains 0s as more 0s = less correllation
# Still, xrd angles seem to have a very low correlation with thickness
```

```python
# While there are not obvious trends, there doees seem to be an outlier with an xrd intensity of 1e8

# Plot boxplot of xrd intensities (11 should suffice) to see if this is a recurring outlier
xrd_intensity_11_col = ["xrd_intensity_"+str(i*100) for i in range(19, 53, 3)]
```

```python
items_per_row = 2
fig, axes = plt.subplots(math.ceil(len(xrd_intensity_11_col)/items_per_row), items_per_row, figsize=(12, len(xrd_intensity_11_col)*2.5))
axes = axes.flatten()

for i, col in enumerate(xrd_intensity_11_col):
    skew_val = df[col].skew()
    sns.boxplot(x=df[col], ax=axes[i])
    axes[i].set_title(f"Boxplot of {col} (Skew: {skew_val:.2f})")

plt.subplots_adjust(hspace=0.3, wspace=0.2) 
plt.show()
```

```python
# max is significantly higher than mean
df_features[xrd_intensity_11_col].describe()
```

```python
# Use a log scaled boxplot for this wide array of values
items_per_row = 2
fig, axes = plt.subplots(math.ceil(len(xrd_intensity_11_col)/items_per_row), items_per_row, figsize=(12, len(xrd_intensity_11_col)*2.5))
axes = axes.flatten()

for i, col in enumerate(xrd_intensity_11_col):
    skew_val = df[col].skew()
    sns.boxplot(x=np.log10(df[col]+1), ax=axes[i])
    axes[i].set_title(f"Log-scaled boxplot of {col} (Skew: {skew_val:.2f})")

plt.subplots_adjust(hspace=0.3, wspace=0.2) 
plt.show()
```

```python
store_outliers = []

for intensity in xrd_intensity_11_col:
    x = np.log10(df_features[intensity] + 1)
    q1, q3 = x.quantile([0.25, 0.75])
    iqr = q3 - q1
    
    outlier_mask = x > q3 + 2.8 * iqr
    outliers = df.loc[outlier_mask]
    store_outliers.append(outliers[['sample_id', 'library_id', intensity]])

store_outliers
```
